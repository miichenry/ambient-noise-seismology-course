**Instructions:**

> Run the cell below first to initialize the quiz engine, then run each section in order.

```python
import time
from IPython.display import display, Markdown, clear_output

score = 0
total = 0
results = []
categories = {}
start_time = time.time()

def ask_mc(question, options, correct_letter, explanation="", category="General"):
    """Multiple choice question."""
    global score, total
    total += 1
    print(f"\n{'='*60}")
    print(f"Question {total}: {question}")
    print(f"{'='*60}")
    for opt in options:
        print(f"  {opt}")
    print()
    ans = input("Your answer (A/B/C/D): ").strip().upper()
    correct = ans == correct_letter.upper()
    if correct:
        score += 1
        print(f"\u2705 Correct!")
    else:
        print(f"\u274c Wrong. The answer is {correct_letter}.")
    results.append((total, correct, category))
    categories.setdefault(category, []).append(correct)
    if explanation:
        print(f"   {explanation}")
    print(f"   Score so far: {score}/{total}")

def ask_fill(question, accepted_answers, explanation="", category="General"):
    """Fill-in-the-blank question."""
    global score, total
    total += 1
    print(f"\n{'='*60}")
    print(f"Question {total}: {question}")
    print(f"{'='*60}")
    ans = input("Your answer: ").strip().lower()
    correct = any(a.lower() in ans for a in accepted_answers)
    if correct:
        score += 1
        print(f"\u2705 Correct!")
    else:
        print(f"\u274c Wrong. Expected: {accepted_answers[0]}")
    results.append((total, correct, category))
    categories.setdefault(category, []).append(correct)
    if explanation:
        print(f"   {explanation}")
    print(f"   Score so far: {score}/{total}")

def ask_tf(question, correct_bool, explanation="", category="General"):
    """True/False question."""
    global score, total
    total += 1
    print(f"\n{'='*60}")
    print(f"Question {total} (True/False): {question}")
    print(f"{'='*60}")
    ans = input("Your answer (T/F): ").strip().upper()
    correct_letter = "T" if correct_bool else "F"
    correct = ans == correct_letter
    if correct:
        score += 1
        print(f"\u2705 Correct!")
    else:
        print(f"\u274c Wrong. The answer is {correct_letter}.")
    results.append((total, correct, category))
    categories.setdefault(category, []).append(correct)
    if explanation:
        print(f"   {explanation}")
    print(f"   Score so far: {score}/{total}")

print("Quiz engine ready. Run the next cells to begin!")
print(f"Timer started at {time.strftime('%H:%M:%S')}")
print("You have 30 minutes for 40 questions. Good luck!")
```
